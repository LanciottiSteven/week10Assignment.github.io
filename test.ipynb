```python
import pandas as pd
import altair as alt
from vega_datasets import data
import json, urllib.request
import numpy as np
```

```python
json_url = data.earthquakes.url
with urllib.request.urlopen(json_url) as f:
    raw = json.load(f)
```

```python
raw['features']
```

```python
df = pd.json_normalize(raw['features'])
df
```

```python
df.columns
```

```python
long_ = []
lat_ = []
for i in df['geometry.coordinates']:
    long_.append(i[0])
    lat_.append(i[1])
df['Lat'] = lat_
df["Long"] = long_
```

```python
df["date_str"] = pd.to_datetime(df["properties.time"])
```

```python
df["date_str"]
```

```python
df['magnitude'] = df['properties.mag'].astype(float)
```
